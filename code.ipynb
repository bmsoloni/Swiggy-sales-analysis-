#importing required modules


```python

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/swiggy_sales_1_year - swiggy_sales_1_year.csv")
display(df)
```

#Missing values


```python


print(df.isnull().sum())

df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum())

```

#Duplicate Records


```python


print("Duplicate records:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate records after cleaning:", df.duplicated().sum())
```

#Incorrect Data Types


```python


df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Item Price"] = pd.to_numeric(
    df["Item Price"], errors="coerce"
)

df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

print(df.dtypes)

```

# Inconsistent Categories

```python


print("Before:")
print(df["Category"].unique())

df["Category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
    .str.title()
)

print("\nAfter:")
print(df["Category"].unique())

```

#Abnormal Values

```python


print("Negative prices:")
print(df[df["Item Price"] < 0])

print("Zero quantity:")
print(df[df["Quantity"] == 0])

```
