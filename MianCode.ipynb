#importing required modules


```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/swiggy_sales_1_year - swiggy_sales_1_year.csv")
```

#1.Which product generates maximum revenue?

```python


df["Revenue"] = df["Net Sales"]

print(
    df.groupby("Item Name")["Revenue"]
      .sum()
      .idxmax())

```

#3. Who are the top 10 customers?

```python


top_10_customers = (
    df.groupby("Order ID")["Revenue"]
      .sum()
      .nlargest(10)
)

print(top_10_customers)

```

#4.Which category has the highest average order value?


```python


# Create Revenue column
df["Revenue"] = df["Net Sales"]

# Category with highest average order value
highest_avg_category = (
    df.groupby("Category")["Revenue"]
      .mean()
      .idxmax()
)

print("Category with highest average order value:", highest_avg_category)

```

#5.Missing values


```python


print(df.isnull().sum())

df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum())

```

#6.Duplicate Records


```python


print("Duplicate records:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate records after cleaning:", df.duplicated().sum())
```

#7.Incorrect Data Types


```python


df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Item Price"] = pd.to_numeric(
    df["Item Price"], errors="coerce"
)

df["Quantity"] = pd.to_numeric(
    df["Quantity"], errors="coerce"
)

print(df.dtypes)

```

#8.Inconsistent Categories

```python


print("Before:")
print(df["Category"].unique())

df["Category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
    .str.title()
)

print("\nAfter:")
print(df["Category"].unique())

```

#9.Abnormal Values

```python


print("Negative prices:")
print(df[df["Item Price"] < 0])

print("Zero quantity:")
print(df[df["Quantity"] == 0])

```
